```python
def is_valid_score(score):
  """ตรวจสอบว่าคะแนนอยู่ในช่วง 0 ถึง 100 หรือไม่

  คืนค่า True หากถูกต้อง และ False หากไม่ถูกต้อง
  """
  return isinstance(score, (int, float)) and 0 <= score <= 100


def to_grade(score):
  """แปลงคะแนนตัวเลขเป็นเกรด (A, B, C, F)

  โดย A ตั้งแต่ 80 ขึ้นไป, B ตั้งแต่ 70 ถึง 79, C ตั้งแต่ 60 ถึง 69
  และต่ำกว่า 60 เป็น F หากคะแนนไม่ถูกต้องจะคืนค่าเป็นข้อความแจ้งเตือน
  """
  if not is_valid_score(score):
    return "คะแนนไม่ถูกต้อง"

  if score >= 80:
    return "A"
  elif score >= 70:
    return "B"
  elif score >= 60:
    return "C"
  else:
    return "F"


def print_result(student_id, score):
  """แสดงผลรายงานคะแนนและเกรดของนักศึกษา

  หากคะแนนไม่ถูกต้องจะแสดงข้อความแจ้งเตือนแทนเกรด
  """
  grade = to_grade(score)
  if grade == "คะแนนไม่ถูกต้อง":
    print(
        f"รหัสนักศึกษา: {student_id} | คะแนน: {score} | แจ้งเตือน: คะแนนไม่ถูกต้อง"
        " (ต้องอยู่ระหว่าง 0 ถึง 100)"
    )
  else:
    print(
        f"รหัสนักศึกษา: {student_id} | คะแนน: {score} | เกรดที่ได้รับ: {grade}"
    )
```

```python
def safe_average(total, count):
  """คืนค่าเฉลี่ยปัดทศนิยม 2 ตำแหน่ง และคืน None เมื่อ count ไม่เป็นบวก"""
  if count <= 0:
    return None
  return round(total / count, 2)


# --- โปรแกรมหลัก ---
# รับค่า PM2.5 จำนวน 5 วันด้วยตัวแปร 5 ตัว
# (หากไม่มีข้อมูลหรือต้องการข้าม สามารถกด Enter หรือใส่ค่าติดลบได้)
print("กรุณากรอกค่า PM2.5 ทั้ง 5 วัน (กดข้ามหรือใส่ค่าติดลบหากไม่มีข้อมูล):")
raw_d1 = input("วันที่ 1: ")
raw_d2 = input("วันที่ 2: ")
raw_d3 = input("วันที่ 3: ")
raw_d4 = input("วันที่ 4: ")
raw_d5 = input("วันที่ 5: ")


# ฟังก์ชันแปลงข้อมูลและตรวจสอบความถูกต้องของค่า PM2.5
def parse_pm25(val_str):
  try:
    val = float(val_str)
    if val < 0:
      return None  # ค่าติดลบถือว่าไม่มีข้อมูลที่ใช้ได้
    return val
  except ValueError:
    return None  # หากกรอกค่าที่ไม่ใช่ตัวเลข


# เก็บค่าลงในตัวแปรทั้ง 5 ตัว
day1 = parse_pm25(raw_d1)
day2 = parse_pm25(raw_d2)
day3 = parse_pm25(raw_d3)
day4 = parse_pm25(raw_d4)
day5 = parse_pm25(raw_d5)

# ตัวสะสมหาผลรวม (total) และจำนวน (count)
total = 0
count = 0

# ตรวจสอบและสะสมข้อมูลจากตัวแปรทั้ง 5 ตัว
for d in [day1, day2, day3, day4, day5]:
  if d is not None:
    total += d
    count += 1

# เรียกใช้งาน safe_average เพื่อคำนวณและแสดงค่าเฉลี่ย
average = safe_average(total, count)

if average is not None:
  print(f"\nค่าเฉลี่ย PM2.5 (จากข้อมูลที่ใช้ได้ {count} วัน): {average}")
else:
  print("\nไม่มีข้อมูลที่ใช้ได้เลย (count ไม่เป็นบวก)")
```
